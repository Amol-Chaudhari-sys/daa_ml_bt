```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier


```

```python
from sklearn.metrics import confusion_matrix, accuracy_score , classification_report
from sklearn.svm import SVC 
```

```python
data = pd.read_csv("emails.csv")
```

```python
data.head (5)
```

```python
print (data.shape)
print ("__________________________")
print (data.info())
print ("__________________________")
print (data.isnull().sum())
```

```python
x= data.drop("Prediction", axis = 1)
y = data["Prediction"]
x= x.drop("Email No.", axis = 1 )
```

```python
x.shape, y.shape
```

```python
x_train , x_test , y_train , y_test  = train_test_split(x, y, random_state= 42 , stratify= y, test_size = 0.2 )
```

```python
scaler = StandardScaler()
x_train= scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
model = KNeighborsClassifier()
model.fit(x_train, y_train)
predect = model.predict(x_test)

accuricy = accuracy_score(predect , y_test)
print (accuricy)

classreport = classification_report(predect , y_test)
print (classreport)
conmat= confusion_matrix(predect , y_test)
conmat
```

```python
svm = SVC(kernel='linear')
svm.fit(x_train , y_train)
svm_predect = svm.predict(x_test)

acc_score_svm = accuracy_score(svm_predect, y_test)
acc_score_svm
```

```python
conf_mat_svm = confusion_matrix(svm_predect, y_test)
print (conf_mat_svm)

report_svm = classification_report(svm_predect, y_test)

print (report_svm )
```

```python
knn_acc = accuricy*100

print ("accuricy of  KNN " , knn_acc )
svm_acc = acc_score_svm*100

print ("accoricy of svm :", svm_acc)
if knn_acc < svm_acc:
    print ("SVM is better ")
elif knn_acc < svm_acc:
    print ("KNN is better ")
else :
    print ("both are equal ")
    
```

```python
plt.bar(["knn algo " , "Svm algo "],[knn_acc, svm_acc]  )
plt.xlabel("Algorithms" )
plt.ylabel("accuricy ")
```

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns


from sklearn.metrics import confusion_matrix , classification_report, accuracy_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.svm import SVC







```

```python
data = pd .read_csv("emails.csv")
data.head ()
```

```python
data= data.drop(["Email No."], axis = 1)
```

```python
x = data.drop ("Prediction", axis = 1 )
y= data["Prediction"]
```

```python
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2 , random_state=42)
```

```python
sc = StandardScaler()
x_train =sc.fit_transform(x_train)
x_test= sc.transform(x_test) 

```

```python
knn = KNeighborsClassifier()
knn.fit(x_train, y_train )
knn_prdect =knn.predict(x_test)
accuracy_score(knn_prdect , y_test)
```

```python
svm  = SVC(kernel="linear")
svm.fit(x_train, y_train )
svm_prdect =svm.predict(x_test)
accuracy_score(svm_prdect , y_test)
```
