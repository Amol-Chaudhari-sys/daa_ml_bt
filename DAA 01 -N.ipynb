```python

import heapq

# Node class for Huffman Tree
class Node:
    def __init__(self, char, freq):
        self.char = char
        self.freq = freq
        self.left = None
        self.right = None

    # Compare nodes based on frequency
    def __lt__(self, other):
        return self.freq < other.freq


# Function to generate Huffman codes
def generate_codes(root, code, huffman_codes):

    if root is None:
        return

    # If leaf node is found
    if root.char is not None:
        huffman_codes[root.char] = code if code else "0"
        return

    generate_codes(root.left, code + "0", huffman_codes)
    generate_codes(root.right, code + "1", huffman_codes)


# Huffman Encoding function
def huffman_encoding(characters, frequencies):

    heap = []

    # Insert nodes into min heap
    for char, freq in zip(characters, frequencies):
        heapq.heappush(heap, Node(char, freq))

    # Build Huffman Tree
    while len(heap) > 1:

        left = heapq.heappop(heap)
        right = heapq.heappop(heap)

        new_node = Node(None, left.freq + right.freq)
        new_node.left = left
        new_node.right = right

        heapq.heappush(heap, new_node)

    # Generate codes
    root = heap[0]
    huffman_codes = {}

    generate_codes(root, "", huffman_codes)

    return huffman_codes


# Main program
n = int(input("Enter number of characters: "))

characters = []
frequencies = []

for i in range(n):
    char = input("Enter character: ")
    freq = int(input("Enter frequency: "))

    characters.append(char)
    frequencies.append(freq)

codes = huffman_encoding(characters, frequencies)

print("\nHuffman Codes:")

for char, code in codes.items():
    print(char, ":", code)
```

```python

# Fractional Knapsack using Greedy Method

def fractional_knapsack(capacity, items):

    # Sort items by profit/weight ratio
    items.sort(key=lambda x: x[1] / x[0], reverse=True)

    total_profit = 0

    print("\nSelected Items:")

    for weight, profit in items:

        if capacity == 0:
            break

        # Take the complete item
        if weight <= capacity:
            capacity -= weight
            total_profit += profit

            print("Weight:", weight, "Profit:", profit)

        # Take a fraction of the item
        else:
            fraction = capacity / weight
            total_profit += profit * fraction

            print("Weight:", capacity,
                  "Fraction:", round(fraction, 2),
                  "Profit:", round(profit * fraction, 2))

            capacity = 0

    return total_profit


# Main Program
n = int(input("Enter number of items: "))

items = []

for i in range(n):
    weight = float(input("Enter weight of item: "))
    profit = float(input("Enter profit of item: "))

    items.append((weight, profit))

capacity = float(input("Enter knapsack capacity: "))

max_profit = fractional_knapsack(capacity, items)

print("\nMaximum Profit =", round(max_profit, 2))
```

```python
# 0/1 Knapsack using Dynamic Programming

def knapsack(weights, profits, capacity):

    n = len(weights)

    # Create DP table
    dp = [[0 for x in range(capacity + 1)]
          for x in range(n + 1)]

    # Fill DP table
    for i in range(1, n + 1):
        for w in range(capacity + 1):

            if weights[i - 1] <= w:

                dp[i][w] = max(
                    profits[i - 1] + dp[i - 1][w - weights[i - 1]],
                    dp[i - 1][w]
                )

            else:
                dp[i][w] = dp[i - 1][w]

    # Find selected items
    selected = []
    w = capacity

    for i in range(n, 0, -1):
        if dp[i][w] != dp[i - 1][w]:
            selected.append(i)
            w -= weights[i - 1]

    return dp[n][capacity], selected


# Main Program

# n = int(input("Enter number of items: "))

weights = [2,1,3,2]
profits = [12, 10,20,15]
# weights = []
# profits = []

# for i in range(n):
#     weight = int(input("Enter weight of item: "))
#     profit = int(input("Enter profit of item: "))

#     weights.append(weight)
#     profits.append(profit)

# capacity = int(input("Enter knapsack capacity: "))
capacity = 5

max_profit, selected_items = knapsack(
    weights, profits, capacity
)

print("\nMaximum Profit =", max_profit)

print("Selected Items:")

for item in reversed(selected_items):
    print("Item", item)
```

# N - Queen 

```python

# N-Queens Problem using Backtracking

def print_board(board, n):
    print("\nFinal N-Queens Matrix:")
    for i in range(n):
        for j in range(n):
            print(board[i][j], end=" ")
        print()


# Check whether queen can be placed safely
def is_safe(board, row, col, n):

    # Check same column
    for i in range(row):
        if board[i][col] == 1:
            return False

    # Check upper-left diagonal
    i = row - 1
    j = col - 1

    while i >= 0 and j >= 0:
        if board[i][j] == 1:
            return False
        i -= 1
        j -= 1

    # Check upper-right diagonal
    i = row - 1
    j = col + 1

    while i >= 0 and j < n:
        if board[i][j] == 1:
            return False
        i -= 1
        j += 1

    return True


# Backtracking function
def solve_n_queens(board, row, n):

    # All queens are placed
    if row == n:
        return True

    # Try each column
    for col in range(n):

        if is_safe(board, row, col, n):

            # Place queen
            board[row][col] = 1

            # Move to next row
            if solve_n_queens(board, row + 1, n):
                return True

            # Backtrack
            board[row][col] = 0

    return False


# Main Program
n = int(input("Enter value of N: "))

if n < 1:
    print("Enter a positive value of N.")

else:
    board = [[0 for j in range(n)] for i in range(n)]

    # Place first queen at (0, 0)
    board[0][0] = 1

    if solve_n_queens(board, 1, n):
        print_board(board, n)
    else:
        print("Solution does not exist with first queen at (0, 0).")
```

# matrix multiplication 

```python

import threading
import time

# Normal Matrix Multiplication
def normal_multiplication(A, B):

    rows_A = len(A)
    cols_A = len(A[0])
    cols_B = len(B[0])

    C = [[0 for j in range(cols_B)]
         for i in range(rows_A)]

    for i in range(rows_A):
        for j in range(cols_B):
            for k in range(cols_A):
                C[i][j] += A[i][k] * B[k][j]

    return C


# Function to multiply one row using a thread
def multiply_row(A, B, C, row):

    cols_A = len(A[0])
    cols_B = len(B[0])

    for j in range(cols_B):
        for k in range(cols_A):
            C[row][j] += A[row][k] * B[k][j]


# Multithreaded Matrix Multiplication
def threaded_multiplication(A, B):

    rows_A = len(A)
    cols_B = len(B[0])

    C = [[0 for j in range(cols_B)]
         for i in range(rows_A)]

    threads = []

    # Create one thread for each row
    for i in range(rows_A):

        t = threading.Thread(
            target=multiply_row,
            args=(A, B, C, i)
        )

        threads.append(t)
        t.start()

    # Wait for all threads
    for t in threads:
        t.join()

    return C

# Main Program
rows_A = int(input("Enter rows of Matrix A: "))
cols_A = int(input("Enter columns of Matrix A: "))

rows_B = int(input("Enter rows of Matrix B: "))
cols_B = int(input("Enter columns of Matrix B: "))

if rows_A <= 0 or cols_A <= 0 or rows_B <= 0 or cols_B <= 0:
    print("Matrix dimensions must be positive.")

elif cols_A != rows_B:
    print("Matrix multiplication is not possible.")

else:

    A = []
    B = []

    print("\nEnter elements of Matrix A:")
    for i in range(rows_A):
        row = list(map(int, input().split()))
        A.append(row)

    print("\nEnter elements of Matrix B:")
    for i in range(rows_B):
        row = list(map(int, input().split()))
        B.append(row)
    # Normal multiplication
    start = time.perf_counter()
    C1 = normal_multiplication(A, B)
    normal_time = time.perf_counter() - start

    # Multithreaded multiplication
    start = time.perf_counter()
    C2 = threaded_multiplication(A, B)
    threaded_time = time.perf_counter() - start

    print("\nNormal Matrix Multiplication:")
    for row in C1:
        print(row)

    print("\nMultithreaded Matrix Multiplication:")
    for row in C2:
        print(row)

    print("\nPerformance Analysis:")
    print("Normal Time:", normal_time, "seconds")
    print("Threaded Time:", threaded_time, "seconds")

    if C1 == C2:
        print("\nBoth results are identical.")
```
