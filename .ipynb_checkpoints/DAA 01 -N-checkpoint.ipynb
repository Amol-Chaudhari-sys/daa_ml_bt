
# Prectical 2 huffman endocing 

```python

import heapq

# Node class for Huffman Tree
class Node:
    def __init__(self, char, freq):
        self.char = char
        self.freq = freq
        self.left = None
        self.right = None

    # Compare nodes based on frequency
    def __lt__(self, other):
        return self.freq < other.freq


# Function to generate Huffman codes
def generate_codes(root, code, huffman_codes):

    if root is None:
        return

    # If leaf node is found
    if root.char is not None:
        huffman_codes[root.char] = code if code else "0"
        return

    generate_codes(root.left, code + "0", huffman_codes)
    generate_codes(root.right, code + "1", huffman_codes)


# Huffman Encoding function
def huffman_encoding(characters, frequencies):

    heap = []

    # Insert nodes into min heap
    for char, freq in zip(characters, frequencies):
        heapq.heappush(heap, Node(char, freq))

    # Build Huffman Tree
    while len(heap) > 1:

        left = heapq.heappop(heap)
        right = heapq.heappop(heap)

        new_node = Node(None, left.freq + right.freq)
        new_node.left = left
        new_node.right = right

        heapq.heappush(heap, new_node)

    # Generate codes
    root = heap[0]
    huffman_codes = {}

    generate_codes(root, "", huffman_codes)

    return huffman_codes


# Main program
n = int(input("Enter number of characters: "))

characters = []
frequencies = []

for i in range(n):
    char = input("Enter character: ")
    freq = int(input("Enter frequency: "))

    characters.append(char)
    frequencies.append(freq)

codes = huffman_encoding(characters, frequencies)

print("\nHuffman Codes:")

for char, code in codes.items():
    print(char, ":", code)
```

```python

# Fractional Knapsack using Greedy Method

def fractional_knapsack(capacity, items):

    # Sort items by profit/weight ratio
    items.sort(key=lambda x: x[1] / x[0], reverse=True)

    total_profit = 0

    print("\nSelected Items:")

    for weight, profit in items:

        if capacity == 0:
            break

        # Take the complete item
        if weight <= capacity:
            capacity -= weight
            total_profit += profit

            print("Weight:", weight, "Profit:", profit)

        # Take a fraction of the item
        else:
            fraction = capacity / weight
            total_profit += profit * fraction

            print("Weight:", capacity,
                  "Fraction:", round(fraction, 2),
                  "Profit:", round(profit * fraction, 2))

            capacity = 0

    return total_profit


# Main Program
n = int(input("Enter number of items: "))

items = []

for i in range(n):
    weight = float(input("Enter weight of item: "))
    profit = float(input("Enter profit of item: "))

    items.append((weight, profit))

capacity = float(input("Enter knapsack capacity: "))

max_profit = fractional_knapsack(capacity, items)

print("\nMaximum Profit =", round(max_profit, 2))
```
