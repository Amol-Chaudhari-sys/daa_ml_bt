```python

// SPDX-License-Identifier: MIT
pragma solidity ^0.8.20;

contract BankAccount {

    // Store balance of each customer
    mapping(address => uint) public balance;

    // Deposit money
    function deposit() public payable {
        require(msg.value > 0, "Enter some amount");

        balance[msg.sender] += msg.value;
    }

    // Withdraw money
    function withdraw(uint amountInEther) public {
        uint amount = amountInEther * 1 ether;

        require(amount > 0, "Amount must be greater than zero");
        require(balance[msg.sender] >= amount, "Insufficient balance");

        balance[msg.sender] -= amount;

        (bool success, ) = payable(msg.sender).call{value: amount}("");
        require(success, "Transfer failed");
    }

    // Show account balance
    function showBalance() public view returns (uint) {
        return balance[msg.sender];
    }
}
```

```python

// SPDX-License-Identifier: MIT
pragma solidity ^0.8.20;

contract BankAccount {

    // Store balance of each customer
    mapping(address => uint) balance;

    // Deposit money
    function deposit(uint256 amount ) public{
        require(amount > 0, "Enter some amount");

        balance[msg.sender] = balance[msg.sender]+ amount;
    }

    // Withdraw money
    function withdraw(uint amount) public {
        require(amount > 0, "Amount must be greater than zero");
        require(balance[msg.sender] >= amount, "Insufficient balance");
        balance[msg.sender] = balance[msg.sender]- amount;
    }

    // Show account balance
    function showBalance() public view returns (uint256) {
        return balance[msg.sender];
    }
}
```

```python

// SPDX-License-Identifier: MIT
pragma solidity ^0.8.20;

contract StudentData {

    // Structure to store student information
    struct Student {
        uint rollNo;
        string name;
        uint marks;
    }

    // Array to store students
    Student[] public students;

    // Function to add student data
    function addStudent(
        uint _rollNo,
        string memory _name,
        uint _marks
    ) public {
        students.push(Student(_rollNo, _name, _marks));
    }

    // Function to get student data
    function getStudent(uint index)
        public
        view
        returns (uint, string memory, uint)
    {
        require(index < students.length, "Invalid index");

        Student memory s = students[index];

        return (s.rollNo, s.name, s.marks);
    }

    // Function to get total number of students
    function getTotalStudents() public view returns (uint) {
        return students.length;
    }

    // Fallback function
    fallback() external payable {
    }

    // Function to check contract balance
    function getBalance() public view returns (uint) {
        return address(this).balance;
    }
}
```
